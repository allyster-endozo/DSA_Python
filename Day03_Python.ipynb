```python
# Retrieve the amount paid > 1500 and payment method via GCash and PayMaya
# Sort the amount paid from highest to lowest
# Calculate the amount paid (total, average, minimum, and maximum)
# Group the total amount paid by payment method
# Write at least three findings in plain language

import numpy as np
import pandas as pd
```

```python
payments = pd.read_csv('payments.csv')
```

```python
payments
```

```python
# Retrieve the amount paid > 1500 and payment method via GCash and PayMaya
# Sort the amount paid from highest to lowest

payments_sorted = payments[
    (pd.to_numeric(payments['amount_paid'], errors = 'coerce') > 1500) & 
    ((payments['payment_method'] == 'GCash') | 
    (payments['payment_method'] == 'PayMaya'))
].sort_values('amount_paid', ascending = False)
```

```python
payments_sorted
```

```python
# Calculate the amount paid (total, average, minimum, and maximum)
# Group the total amount paid by payment method

payments_sorted['amount_paid'] = pd.to_numeric(
    payments_sorted['amount_paid'],
    errors = 'coerce'
)

payments_aggregation = (
    payments_sorted.groupby('payment_method')
        .agg(
            amount_paid_sum = ('amount_paid', 'sum'),
            amount_paid_mean = ('amount_paid', 'mean'),
            amount_paid_min = ('amount_paid', 'min'),
            amount_paid_max = ('amount_paid', 'max')
        )
        .reset_index()
)
```

```python
payments_aggregation
```

```python
# Write at least three findings in plain language

amount_paid_sum_GCash = payments_aggregation.iloc[0]['amount_paid_sum']
amount_paid_sum_PayMaya = payments_aggregation.iloc[1]['amount_paid_sum']

amount_paid_mean_GCash = payments_aggregation.iloc[0]['amount_paid_mean']
amount_paid_mean_PayMaya = payments_aggregation.iloc[1]['amount_paid_mean']

amount_paid_min_GCash = payments_aggregation.iloc[0]['amount_paid_min']
amount_paid_min_PayMaya = payments_aggregation.iloc[1]['amount_paid_min']

amount_paid_max_GCash = payments_aggregation.iloc[0]['amount_paid_max']
amount_paid_max_PayMaya = payments_aggregation.iloc[1]['amount_paid_max']

print(f"In terms of the amount paid exceeding PHP 1,500.00:", end = "\n")

print(f"• PayMaya registered a sum amounting to PHP {amount_paid_sum_PayMaya:,.2f} — exceeding that of GCash amounting to PHP {amount_paid_sum_GCash:,.2f}.", end = "\n")

print(f"• GCash registered a mean amounting to PHP {amount_paid_mean_GCash:,.2f} — exceeding that of PayMaya amounting to PHP {amount_paid_mean_PayMaya:,.2f}.", end = "\n")

print(f"• PayMaya received payments ranging from PHP {amount_paid_min_PayMaya:,.2f} to {amount_paid_max_PayMaya:,.2f} — varying more than those of GCash ranging from PHP {amount_paid_min_GCash:,.2f} to {amount_paid_max_GCash:,.2f}.")
```
