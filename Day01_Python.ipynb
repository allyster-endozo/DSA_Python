```python
if score >= 90:
    rating = "Excellent"
elif score >= 80 and score < 90:
    rating = "Very Good"
elif score >= 70 and score < 80:
    rating = "Good"
else:
    rating = "Needs Improvement"

print(rating)
```

```python
product = "Laptop"
qty = 4
unit_price = 35000
unit_cost = 28000

revenue = qty * unit_price
cost = qty * unit_cost
profit = revenue - cost
margin = profit / revenue * 100
```

```python
print(margin)
```

```python
if margin >= 20:
    condition = "healthy"
else:
    condition = "not healthy"

print(condition)
```

```python
print(f"The profit margin has been found to be {condition} — considering the profit at PHP {profit:,.2f} reaching at least 20% of the revenue at PHP {revenue:,.2f}.")
```
