```python
# combining loops and conditions
sales = [45000, 82000, 120000, 67000, 95000]

high_count = 0

for amount in sales:
    if amount >= 100000:
        label = "Excellent"
        high_count += 1
    elif amount >= 75000:
        label = "Good"
    else:
        label = "Needs review"

print(amount, label)
print(f"Excellent records: {high_count}")
```

```python
# while loops
count = 1

while count <= 5:
    print(f"Attempt {count}")
    count += 1

print("Loop finished")
```

```python
# the break statement ends a while loop
count = 0

while True:
    count += 1
    if count == 3:
        break
        
print(count)
```

```python
# the continue statement enables skipping
count = 0

while count < 5:
    count += 1
    if count == 3:
        continue
    print(count)
```

```python
def summarize(values):
    total = sum(values)
    average = total / len(values)
    return total, average
 
result = summarize([10, 20, 30])
print(result)
```

```python
# parameters, arguments, and return values
def calculate_margin(revenue, cost): # calculate_margin is a function
    profit = revenue - cost # revenue and cost are the parameters
    margin = profit / revenue * 100
    return margin

margin = calculate_margin(150000, 120000) # 150000, 120000 constitutes the argument

if margin >= 20:
    print("Healthy margin")
```

```python
sales = [45000, 82000, 120000, 67000, 95000]
target = 75000
# Your task:
# 1. Calculate total and average
# 2. Classify each value
# 3. Count values >= target
# 4. Use at least one function
```

```python
def is_target_met(amount, target = 75000):
    return amount >= target

for index, amount in enumerate(sales):
    if is_target_met(amount):
        print(f"Value {index + 1}: PHP {amount:,.2f} (On-target)")
    else:
        print(f"Value {index + 1}: PHP {amount:,.2f} (Off-target)")

print(f"", end="\n")

average = sum(sales) / len(sales)

print(f"Total sales: PHP {sum(sales):,.2f}")
print(f"Average sales: PHP {average:,.2f}")
print(f"On-target count: {sum(x > 75000 for x in sales)}")
```
